```python
%%writefile monkey.pl

% Monkey Banana Problem

% Initial state:
% Monkey is at door, box is at window,
% banana is at middle.

% Monkey walks to the box
move(state(door, floor, window, hanging),
     state(window, floor, window, hanging)) :-
    write('Monkey walks from door to window.'), nl.

% Monkey pushes box to banana
move(state(window, floor, window, hanging),
     state(middle, floor, middle, hanging)) :-
    write('Monkey pushes the box from window to middle.'), nl.

% Monkey climbs onto the box
move(state(middle, floor, middle, hanging),
     state(middle, onbox, middle, hanging)) :-
    write('Monkey climbs onto the box.'), nl.

% Monkey gets the banana
move(state(middle, onbox, middle, hanging),
     state(middle, onbox, middle, grabbed)) :-
    write('Monkey grabs the banana.'), nl.

% Goal state
solve(state(middle, onbox, middle, grabbed)) :-
    write('Monkey has got the banana!'), nl.

% Solve using the above moves
solve(State) :-
    move(State, NewState),
    solve(NewState).
```
