```python
%%writefile forward.pl

% Forward Chaining Program

:- dynamic fact/1.

% Initial facts
fact(bird).
fact(has_wings).

% Rules
rule(flies) :-
    fact(bird),
    fact(has_wings).

rule(has_feathers) :-
    fact(bird).

% Forward chaining
forward :-
    rule(NewFact),
    \+ fact(NewFact),
    assertz(fact(NewFact)),
    write('New fact derived: '),
    write(NewFact),
    nl,
    forward.

forward :-
    write('Forward chaining completed.'), nl.
```

```python
!swipl -q -s forward.pl -g "forward,halt."
```

```python
!swipl -q -s forward.pl -g "forward,fact(flies),write('flies is TRUE'),nl,halt."
```
