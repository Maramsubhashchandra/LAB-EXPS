```python
# Python program to solve Water Jug Problem using BFS

from collections import deque

def water_jug(capacity1, capacity2, target):
    # Initial state: both jugs are empty
    start = (0, 0)

    # Queue stores state and path
    queue = deque([(start, [])])

    # To avoid visiting the same state repeatedly
    visited = {start}

    while queue:
        (jug1, jug2), path = queue.popleft()

        # Check if target is reached
        if jug1 == target or jug2 == target:
            return path + [(jug1, jug2)]

        # Generate all possible operations
        states = [
            ((capacity1, jug2), "Fill Jug 1"),
            ((jug1, capacity2), "Fill Jug 2"),
            ((0, jug2), "Empty Jug 1"),
            ((jug1, 0), "Empty Jug 2")
        ]

        # Pour Jug 1 into Jug 2
        amount = min(jug1, capacity2 - jug2)
        states.append(
            ((jug1 - amount, jug2 + amount), "Pour Jug 1 -> Jug 2")
        )

        # Pour Jug 2 into Jug 1
        amount = min(jug2, capacity1 - jug1)
        states.append(
            ((jug1 + amount, jug2 - amount), "Pour Jug 2 -> Jug 1")
        )

        for new_state, action in states:
            if new_state not in visited:
                visited.add(new_state)
                queue.append(
                    (new_state, path + [(jug1, jug2, action)])
                )

    return None


# Get input from the user
capacity1 = int(input("Enter capacity of Jug 1: "))
capacity2 = int(input("Enter capacity of Jug 2: "))
target = int(input("Enter target amount: "))

# Solve the problem
solution = water_jug(capacity1, capacity2, target)

if solution:
    print("\nSolution Steps:")

    for step in solution:
        if len(step) == 2:
            print(f"Jug 1 = {step[0]}, Jug 2 = {step[1]}")
        else:
            print(f"{step[2]}")
            print(f"Jug 1 = {step[0]}, Jug 2 = {step[1]}")
else:
    print("No solution exists.")
```
