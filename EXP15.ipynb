```python
# Python program to solve Vacuum Cleaner Problem

# Initial state of the rooms
room_A = input("Enter status of Room A (Clean/Dirty): ").lower()
room_B = input("Enter status of Room B (Clean/Dirty): ").lower()

# Initial position of vacuum cleaner
position = input("Enter vacuum position (A/B): ").upper()

print("\nInitial State:")
print("Room A:", room_A)
print("Room B:", room_B)
print("Vacuum Position:", position)

# Clean Room A
if position == "A":
    if room_A == "dirty":
        print("\nVacuum cleans Room A.")
        room_A = "clean"

    print("Vacuum moves from Room A to Room B.")
    position = "B"

# Clean Room B
if position == "B":
    if room_B == "dirty":
        print("Vacuum cleans Room B.")
        room_B = "clean"

# Display final state
print("\nFinal State:")
print("Room A:", room_A)
print("Room B:", room_B)
print("Vacuum Position:", position)

if room_A == "clean" and room_B == "clean":
    print("Both rooms are clean!")
```
