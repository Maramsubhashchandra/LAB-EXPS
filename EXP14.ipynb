```python
# Python program to solve Missionaries and Cannibals Problem
# Using Breadth-First Search (BFS)

from collections import deque

# State: (missionaries_left, cannibals_left, boat_position)
# boat_position: 0 = Left, 1 = Right

def is_valid(state):
    m_left, c_left, boat = state

    m_right = 3 - m_left
    c_right = 3 - c_left

    # Check valid number of people
    if not (0 <= m_left <= 3 and 0 <= c_left <= 3):
        return False

    # Missionaries should not be outnumbered on the left bank
    if m_left > 0 and m_left < c_left:
        return False

    # Missionaries should not be outnumbered on the right bank
    if m_right > 0 and m_right < c_right:
        return False

    return True


def get_next_states(state):
    m_left, c_left, boat = state

    # Possible combinations carried by boat
    moves = [
        (1, 0),  # 1 Missionary
        (2, 0),  # 2 Missionaries
        (0, 1),  # 1 Cannibal
        (0, 2),  # 2 Cannibals
        (1, 1)   # 1 Missionary and 1 Cannibal
    ]

    next_states = []

    for m, c in moves:

        if boat == 0:  # Boat on left bank
            new_state = (m_left - m, c_left - c, 1)
        else:          # Boat on right bank
            new_state = (m_left + m, c_left + c, 0)

        if is_valid(new_state):
            next_states.append(new_state)

    return next_states


def solve():
    start = (3, 3, 0)
    goal = (0, 0, 1)

    queue = deque([(start, [start])])
    visited = {start}

    while queue:
        state, path = queue.popleft()

        if state == goal:
            return path

        for next_state in get_next_states(state):
            if next_state not in visited:
                visited.add(next_state)
                queue.append((next_state, path + [next_state]))

    return None


def display_state(state):
    m_left, c_left, boat = state

    m_right = 3 - m_left
    c_right = 3 - c_left

    if boat == 0:
        boat_position = "Left"
    else:
        boat_position = "Right"

    print(
        f"Left: M={m_left}, C={c_left} | "
        f"Right: M={m_right}, C={c_right} | "
        f"Boat: {boat_position}"
    )


# Main program
solution = solve()

if solution:
    print("Solution for Missionaries and Cannibals Problem:\n")

    for i, state in enumerate(solution):
        print(f"Step {i}: ", end="")
        display_state(state)
else:
    print("No solution exists.")
```
