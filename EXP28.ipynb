```python
%%writefile student_teacher.pl

% Student facts
student(rahul).
student(priya).
student(amit).
student(neha).

% Teacher facts
teacher(ravi).
teacher(sita).
teacher(rajesh).

% Subject facts
subject(math).
subject(physics).
subject(computer_science).

% Teacher teaches Subject
teaches(ravi, math).
teaches(sita, physics).
teaches(rajesh, computer_science).

% Student studies Subject
studies(rahul, math).
studies(priya, physics).
studies(amit, computer_science).
studies(neha, math).

% Find the teacher of a student
student_teacher(Student, Teacher) :-
    studies(Student, Subject),
    teaches(Teacher, Subject).
```
