```python
# Python program to demonstrate List Operations

# 1. Nested List
list1 = [10, 20, 30]
list2 = [40, 50, 60]
nested_list = [list1, list2]

print("Nested List:", nested_list)

# 2. Length
print("Length of list1:", len(list1))

# 3. Concatenation
list3 = list1 + list2
print("Concatenated List:", list3)

# 4. Membership
print("Is 20 present in list1?", 20 in list1)
print("Is 100 present in list1?", 100 in list1)

# 5. Iteration
print("Elements using iteration:")
for element in list1:
    print(element)

# 6. Indexing
print("First element:", list1[0])
print("Second element:", list1[1])
print("Last element:", list1[-1])

# 7. Slicing
print("First two elements:", list3[0:2])
print("Elements from index 2 to 4:", list3[2:5])
print("All elements in reverse:", list3[::-1])
```
