```python
# Python program to implement Decision Tree

from sklearn.tree import DecisionTreeClassifier

# Training data
# Features: [Age, Salary]
X = [
    [25, 20000],
    [30, 30000],
    [35, 40000],
    [40, 50000],
    [45, 60000],
    [50, 70000]
]

# Output: 0 = No, 1 = Yes
y = [0, 0, 0, 1, 1, 1]

# Create Decision Tree model
model = DecisionTreeClassifier()

# Train the model
model.fit(X, y)

# Predict for a new person
age = int(input("Enter age: "))
salary = int(input("Enter salary: "))

prediction = model.predict([[age, salary]])

# Display result
if prediction[0] == 1:
    print("Prediction: Yes")
else:
    print("Prediction: No")
```
