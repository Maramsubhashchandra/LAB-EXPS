```python
%%writefile planets.pl

% PLANETS DATABASE

planet(mercury).
planet(venus).
planet(earth).
planet(mars).
planet(jupiter).
planet(saturn).
planet(uranus).
planet(neptune).

% Planet and its position from the Sun
position(mercury, 1).
position(venus, 2).
position(earth, 3).
position(mars, 4).
position(jupiter, 5).
position(saturn, 6).
position(uranus, 7).
position(neptune, 8).

% Planet types
type(mercury, terrestrial).
type(venus, terrestrial).
type(earth, terrestrial).
type(mars, terrestrial).
type(jupiter, gas_giant).
type(saturn, gas_giant).
type(uranus, ice_giant).
type(neptune, ice_giant).

% Find the position of a planet
find_position(Planet, Position) :-
    position(Planet, Position).

% Find the type of a planet
find_type(Planet, Type) :-
    type(Planet, Type).
```

```python
!swipl -q -s planets.pl -g "find_position(earth,P),write('Earth is planet number '),write(P),nl,halt."
```

```python
!swipl -q -s planets.pl -g "find_type(jupiter,T),write('Jupiter is a '),write(T),nl,halt."
```
