```python
%%writefile diet.pl

% Diet suggestion system based on disease

% Disease facts
disease(diabetes).
disease(hypertension).
disease(anemia).
disease(obesity).
disease(gastritis).

% Diet suggestions

diet(diabetes, 'Eat vegetables, whole grains and low-sugar fruits.').
diet(diabetes, 'Avoid sugary drinks, sweets and excess refined carbohydrates.').

diet(hypertension, 'Eat fruits, vegetables and low-sodium foods.').
diet(hypertension, 'Avoid excess salt, processed foods and high-sodium snacks.').

diet(anemia, 'Eat iron-rich foods such as leafy vegetables, beans and fortified cereals.').
diet(anemia, 'Include vitamin-C-rich foods to help iron absorption.').

diet(obesity, 'Eat vegetables, whole grains and other nutrient-dense foods in appropriate portions.').
diet(obesity, 'Limit highly processed foods and sugary drinks.').

diet(gastritis, 'Choose mild foods and eat smaller meals as tolerated.').
diet(gastritis, 'Avoid foods that trigger symptoms, such as very spicy or acidic foods.').

% Rule to suggest diet
suggest_diet(Disease, Diet) :-
    disease(Disease),
    diet(Disease, Diet).
```
