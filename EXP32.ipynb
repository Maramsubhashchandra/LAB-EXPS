```python
%%writefile family.pl

% Gender facts
female(pam).
female(liz).
female(ann).
female(pat).

male(tom).
male(bob).
male(jim).

% Parent facts
parent(pam, liz).
parent(pam, bob).
parent(tom, liz).
parent(tom, bob).

parent(liz, ann).
parent(liz, jim).
parent(bob, pat).

% Mother relation
mother(X, Y) :-
    female(X),
    parent(X, Y).

% Father relation
father(X, Y) :-
    male(X),
    parent(X, Y).

% Grandfather relation
grandfather(X, Y) :-
    male(X),
    parent(X, Z),
    parent(Z, Y).

% Grandmother relation
grandmother(X, Y) :-
    female(X),
    parent(X, Z),
    parent(Z, Y).

% Sister relation
sister(X, Y) :-
    female(X),
    parent(P, X),
    parent(P, Y),
    X \= Y.

% Brother relation
brother(X, Y) :-
    male(X),
    parent(P, X),
    parent(P, Y),
    X \= Y.
```

```python
!swipl -q -s family.pl -g "mother(pam,liz),write('Pam is the mother of Liz.'),nl,halt."
!swipl -q -s family.pl -g "father(tom,bob),write('Tom is the father of Bob.'),nl,halt."
!swipl -q -s family.pl -g "grandfather(tom,ann),write('Tom is the grandfather of Ann.'),nl,halt."
!swipl -q -s family.pl -g "grandmother(pam,ann),write('Pam is the grandmother of Ann.'),nl,halt."
!swipl -q -s family.pl -g "sister(ann,jim),write('Ann is the sister of Jim.'),nl,halt."
!swipl -q -s family.pl -g "brother(jim,ann),write('Jim is the brother of Ann.'),nl,halt."
```
