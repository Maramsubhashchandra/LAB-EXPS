```python
# Python program for Map Coloring using CSP
# Constraint: Adjacent regions must have different colors

# Map represented as a graph
map_graph = {
    'A': ['B', 'C'],
    'B': ['A', 'C', 'D'],
    'C': ['A', 'B', 'D'],
    'D': ['B', 'C', 'E'],
    'E': ['D']
}

# Available colors
colors = ['Red', 'Green', 'Blue']


def is_safe(region, color, assignment):
    # Check all neighboring regions
    for neighbour in map_graph[region]:
        if neighbour in assignment and assignment[neighbour] == color:
            return False

    return True


def solve(assignment):
    # If all regions are colored
    if len(assignment) == len(map_graph):
        return True

    # Select an uncolored region
    for region in map_graph:
        if region not in assignment:
            break

    # Try each available color
    for color in colors:
        if is_safe(region, color, assignment):
            assignment[region] = color

            # Recursive backtracking
            if solve(assignment):
                return True

            # Remove color if it leads to failure
            del assignment[region]

    return False


# Create an empty assignment
assignment = {}

# Solve the CSP
if solve(assignment):
    print("Map Coloring Solution:")
    for region in sorted(assignment):
        print(region, "->", assignment[region])
else:
    print("No solution exists.")
```
