```python
%%writefile birds.pl

% Bird facts
bird(sparrow).
bird(parrot).
bird(eagle).
bird(penguin).
bird(ostrich).
bird(crow).

% Birds that can fly
can_fly(sparrow).
can_fly(parrot).
can_fly(eagle).
can_fly(crow).

% Birds that cannot fly
cannot_fly(penguin).
cannot_fly(ostrich).

% Rule to check whether a particular bird can fly
flies(Bird) :-
    can_fly(Bird),
    write(Bird),
    write(' can fly.'),
    nl.

flies(Bird) :-
    cannot_fly(Bird),
    write(Bird),
    write(' cannot fly.'),
    nl.
```
