```python
%%writefile hanoi.pl

% Towers of Hanoi

hanoi(1, Source, Destination, _) :-
    write('Move disk from '),
    write(Source),
    write(' to '),
    write(Destination),
    nl.

hanoi(N, Source, Destination, Auxiliary) :-
    N > 1,
    N1 is N - 1,

    hanoi(N1, Source, Auxiliary, Destination),

    write('Move disk from '),
    write(Source),
    write(' to '),
    write(Destination),
    nl,

    hanoi(N1, Auxiliary, Destination, Source).
```
