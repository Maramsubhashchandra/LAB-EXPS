```python
# Python program to demonstrate Set Operations

# Create two sets
set1 = {10, 20, 30, 40, 50}
set2 = {30, 40, 50, 60, 70}

print("Set 1:", set1)
print("Set 2:", set2)

# 1. Union
print("Union:", set1.union(set2))

# 2. Intersection
print("Intersection:", set1.intersection(set2))

# 3. Difference
print("Difference (Set1 - Set2):", set1.difference(set2))
print("Difference (Set2 - Set1):", set2.difference(set1))

# 4. Symmetric Difference
print("Symmetric Difference:", set1.symmetric_difference(set2))

# 5. Membership
print("Is 20 present in Set1?", 20 in set1)
print("Is 100 present in Set1?", 100 in set1)

# 6. Subset
set3 = {10, 20}
print("Is Set3 a subset of Set1?", set3.issubset(set1))

# 7. Superset
print("Is Set1 a superset of Set3?", set1.issuperset(set3))
```
