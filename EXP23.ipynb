```python
# Python program to implement Alpha-Beta Pruning
# for Tic-Tac-Toe Game

import math

# Create the board
board = [' ' for _ in range(9)]


def display_board():
    print()
    print(board[0], "|", board[1], "|", board[2])
    print("--+---+--")
    print(board[3], "|", board[4], "|", board[5])
    print("--+---+--")
    print(board[6], "|", board[7], "|", board[8])
    print()


def check_winner(player):
    winning_positions = [
        (0, 1, 2),
        (3, 4, 5),
        (6, 7, 8),
        (0, 3, 6),
        (1, 4, 7),
        (2, 5, 8),
        (0, 4, 8),
        (2, 4, 6)
    ]

    for a, b, c in winning_positions:
        if board[a] == board[b] == board[c] == player:
            return True

    return False


def is_draw():
    return ' ' not in board


def alpha_beta(is_maximizing, alpha, beta):
    # Terminal states
    if check_winner('O'):
        return 1

    if check_winner('X'):
        return -1

    if is_draw():
        return 0

    if is_maximizing:
        # Computer tries to maximize the score
        best_score = -math.inf

        for i in range(9):
            if board[i] == ' ':
                board[i] = 'O'

                score = alpha_beta(False, alpha, beta)

                board[i] = ' '

                best_score = max(best_score, score)
                alpha = max(alpha, best_score)

                # Alpha-Beta pruning
                if beta <= alpha:
                    break

        return best_score

    else:
        # Human tries to minimize the score
        best_score = math.inf

        for i in range(9):
            if board[i] == ' ':
                board[i] = 'X'

                score = alpha_beta(True, alpha, beta)

                board[i] = ' '

                best_score = min(best_score, score)
                beta = min(beta, best_score)

                # Alpha-Beta pruning
                if beta <= alpha:
                    break

        return best_score


def best_move():
    best_score = -math.inf
    move = None

    for i in range(9):
        if board[i] == ' ':
            board[i] = 'O'

            score = alpha_beta(False, -math.inf, math.inf)

            board[i] = ' '

            if score > best_score:
                best_score = score
                move = i

    return move


# Main game
print("TIC-TAC-TOE using Alpha-Beta Pruning")
print("You are X and Computer is O.")

while True:
    display_board()

    # Human move
    try:
        position = int(input("Enter your position (1-9): ")) - 1
    except ValueError:
        print("Enter a valid number.")
        continue

    if position < 0 or position > 8 or board[position] != ' ':
        print("Invalid position!")
        continue

    board[position] = 'X'

    if check_winner('X'):
        display_board()
        print("You win!")
        break

    if is_draw():
        display_board()
        print("Game Draw!")
        break

    # Computer move
    computer_move = best_move()
    board[computer_move] = 'O'

    print("Computer chooses position:", computer_move + 1)

    if check_winner('O'):
        display_board()
        print("Computer wins!")
        break

    if is_draw():
        display_board()
        print("Game Draw!")
        break
```
