```python
# Python program to demonstrate List methods

# Create a list
my_list = [10, 20, 30]

print("Original List:", my_list)

# 1. Add - using insert()
my_list.insert(1, 15)
print("After Add:", my_list)

# 2. Append - adds an element at the end
my_list.append(40)
print("After Append:", my_list)

# 3. Extend - adds multiple elements
my_list.extend([50, 60])
print("After Extend:", my_list)

# 4. Delete - using remove()
my_list.remove(20)
print("After Delete:", my_list)
```
