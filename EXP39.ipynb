```python
%%writefile backward.pl

% Backward Chaining Program

% Facts
bird(sparrow).
bird(parrot).
bird(penguin).

has_wings(sparrow).
has_wings(parrot).

% Rules
flies(X) :-
    bird(X),
    has_wings(X).

can_move(X) :-
    flies(X).

% Backward chaining query
backward_chaining(X) :-
    can_move(X),
    write(X),
    write(' can move.'),
    nl.
```

```python
!swipl -q -s backward.pl -g "backward_chaining(sparrow),halt."
```

```python
!swipl -q -s backward.pl -g "flies(sparrow),write('Sparrow can fly.'),nl,halt."
```
