```python
# Python program to remove punctuations from a string

import string

# Get input string
text = input("Enter a string: ")

# Remove punctuation
result = ""

for char in text:
    if char not in string.punctuation:
        result += char

# Display the result
print("String after removing punctuations:")
print(result)
```
