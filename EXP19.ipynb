```python
# Python program to implement A* Algorithm

import heapq

# Graph with edge costs
graph = {
    'A': {'B': 1, 'C': 4},
    'B': {'D': 2, 'E': 5},
    'C': {'E': 1},
    'D': {'F': 3},
    'E': {'F': 2},
    'F': {}
}

# Heuristic values (estimated cost to goal F)
heuristic = {
    'A': 6,
    'B': 5,
    'C': 3,
    'D': 3,
    'E': 2,
    'F': 0
}


def a_star(start, goal):
    # Priority queue: (f_cost, g_cost, node, path)
    queue = [(heuristic[start], 0, start, [start])]

    # Store the lowest cost found for each node
    cost_so_far = {start: 0}

    while queue:
        f_cost, g_cost, current, path = heapq.heappop(queue)

        # Goal reached
        if current == goal:
            return path, g_cost

        # Explore neighboring nodes
        for neighbour, cost in graph[current].items():

            new_cost = g_cost + cost

            # If a cheaper path is found
            if neighbour not in cost_so_far or \
                    new_cost < cost_so_far[neighbour]:

                cost_so_far[neighbour] = new_cost

                # f(n) = g(n) + h(n)
                f = new_cost + heuristic[neighbour]

                heapq.heappush(
                    queue,
                    (f, new_cost, neighbour, path + [neighbour])
                )

    return None, float('inf')


# Start and goal nodes
start = 'A'
goal = 'F'

# Run A* algorithm
path, cost = a_star(start, goal)

# Display result
if path:
    print("A* Algorithm")
    print("------------")
    print("Shortest Path:", " -> ".join(path))
    print("Total Cost:", cost)
else:
    print("No path found.")
```
