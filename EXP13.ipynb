```python
# Python program to solve Crypt-Arithmetic Problem
# SEND + MORE = MONEY

from itertools import permutations

# Digits available
digits = range(10)

# Try all possible digit combinations
for perm in permutations(digits, 8):

    S, E, N, D, M, O, R, Y = perm

    # S and M cannot be zero
    if S == 0 or M == 0:
        continue

    # Form the numbers
    SEND = 1000*S + 100*E + 10*N + D
    MORE = 1000*M + 100*O + 10*R + E
    MONEY = 10000*M + 1000*O + 100*N + 10*E + Y

    # Check the equation
    if SEND + MORE == MONEY:
        print("Solution Found:")
        print("S =", S)
        print("E =", E)
        print("N =", N)
        print("D =", D)
        print("M =", M)
        print("O =", O)
        print("R =", R)
        print("Y =", Y)

        print("\nSEND =", SEND)
        print("MORE =", MORE)
        print("MONEY =", MONEY)

        break
```
