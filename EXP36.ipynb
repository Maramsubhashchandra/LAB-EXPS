```python
%%writefile bestfirst.pl

% Best First Search in Prolog

% Graph
edge(a, b).
edge(a, c).
edge(b, d).
edge(b, e).
edge(c, f).
edge(c, g).
edge(e, h).
edge(f, h).

% Heuristic values
h(a, 6).
h(b, 4).
h(c, 3).
h(d, 6).
h(e, 2).
h(f, 1).
h(g, 5).
h(h, 0).

% Best First Search
best_first(Start, Goal, Path) :-
    search([[Start]], Goal, RevPath),
    reverse(RevPath, Path).

search([[Goal|Path]|_], Goal, [Goal|Path]).

search([[Current|Path]|Rest], Goal, Solution) :-
    findall(
        [Next,Current|Path],
        (edge(Current, Next),
         \+ member(Next, [Current|Path])),
        NewPaths
    ),
    add_paths(NewPaths, Rest, UpdatedQueue),
    sort_queue(UpdatedQueue, SortedQueue),
    search(SortedQueue, Goal, Solution).

% Add paths to queue
add_paths([], Queue, Queue).

add_paths([Path|Paths], Queue, NewQueue) :-
    add_paths(Paths, [Path|Queue], NewQueue).

% Sort paths according to heuristic of first node
sort_queue(Paths, SortedPaths) :-
    map_list_to_pairs(path_heuristic, Paths, Pairs),
    keysort(Pairs, SortedPairs),
    pairs_values(SortedPairs, SortedPaths).

path_heuristic([Node|_], H) :-
    h(Node, H).
```
