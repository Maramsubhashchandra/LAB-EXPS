```python
%%writefile medical.pl

% Medical Diagnosis System

% Symptoms and diseases

diagnosis(flu) :-
    symptom(fever),
    symptom(cough),
    symptom(body_pain),
    write('Diagnosis: Flu'), nl.

diagnosis(cold) :-
    symptom(cough),
    symptom(runny_nose),
    symptom(sneezing),
    write('Diagnosis: Common Cold'), nl.

diagnosis(malaria) :-
    symptom(fever),
    symptom(chills),
    symptom(headache),
    write('Diagnosis: Malaria'), nl.

diagnosis(diabetes) :-
    symptom(frequent_urination),
    symptom(excessive_thirst),
    write('Diagnosis: Diabetes'), nl.

% Facts about symptoms

symptom(fever).
symptom(cough).
symptom(body_pain).
symptom(runny_nose).
symptom(sneezing).
symptom(chills).
symptom(headache).
symptom(frequent_urination).
symptom(excessive_thirst).
```
